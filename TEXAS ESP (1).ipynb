```python
#Encoding the categorical data
from sklearn.preprocessing import OneHotEncoder
OHE=OneHotEncoder(handle_unknown='ignore',sparse_output=False).set_output(transform='pandas')
```

```python
df.columns
```

```python
oh1=OHE.fit_transform(df[['ETHNICITY','GENDER','STATUS']])
```

```python
df=df.drop(columns=['ETHNICITY','GENDER','STATUS'])
```

```python
df=pd.concat([df,oh1],axis=1)
```

```python
#handling outliers
```

```python
cols = ['MONTHLY', 'ANNUAL']  # numeric columns to check
for col in cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5*IQR
    upper = Q3 + 1.5*IQR
    df = df[(df[col] >= lower) & (df[col] <= upper)]
```

```python
from scipy import stats
z_scores = np.abs(stats.zscore(df['HRLY RATE']))
threshold = 3  # common choice: 3 standard deviations
df_no_outliers = df[z_scores < threshold]
```

```python
z_scores = np.abs(stats.zscore(df['HRS PER WK']))
threshold = 3  # common choice: 3 standard deviations
df_no_outliers = df[z_scores < threshold]
```

```python
#splitting the data
```

```python
x = df.drop('ANNUAL',axis=1)
y = df['ANNUAL']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2, random_state = 42)
```

```python
#scaling the data
from sklearn.preprocessing import MinMaxScaler
scaler=MinMaxScaler()
```

```python
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.transform(x_test)
```

TASK 2

MODEL BUILDING

```python

from sklearn.linear_model import LinearRegression
LR=LinearRegression()
```

```python

model = LinearRegression()
model.fit(x_train_scaled, y_train)
y_pred = model.predict(x_test_scaled)
```

TESTING

```python
from sklearn.metrics import (mean_absolute_error,mean_squared_error,r2_score)
```

```python
#testing data
# Mean Absolute Error
mae = mean_absolute_error(y_test, y_pred)

# Mean Squared Error
mse = mean_squared_error(y_test, y_pred)

# Root Mean Squared Error
rmse = np.sqrt(mse)

# R-squared
r2 = r2_score(y_test, y_pred)

# Print results
print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R²:", r2)
```

```python
y_train_predict=model.predict(x_train_scaled)
```

```python
#traing data
mae = mean_absolute_error(y_train, y_train_predict)
mse = mean_squared_error(y_train, y_train_predict)
rmse = np.sqrt(mse)
r2 = r2_score(y_train, y_train_predict)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R²:", r2)
```
